# 파이썬 기초 및 자료형
- R, Python, SAS등 통계분석을 하기 위한 프로그램은 여러가지 존재한다. 하지만, 이들을 사용하기 위해서는 컴퓨터에게 명령을 할줄 알아야 한다. 즉, 명령을 하기위한 언어를 익혀야 한다. 그 중, 파이썬에서는 가장 기본이 되는 자료형에 대해서 학습한다.

## 1. Number(숫자형)
- Number(숫자형)은 정수, 실수 등 숫자를 표현하는 자료형이다.
    - Integer(정수형, int) : 정수를 표현하는 자료형이다.
    - Float(실수형, float) : 소수점이 포함된 숫자를 표현하는 자료형이다.
- type() : 값의 자료형을 확인하는 내장 함수이다.
- print() : 값을 화면에 출력하는 내장 함수이다.
- 내장 함수란 파이썬에서 기본적으로 제공하는 함수를 의미한다.

```python
x = 123
print(x)
print(type(x))
```

```python
x = -1.23
print(x)
print(type(x))
```

```python
x = 1.23e-5
print(x)
print(type(x))
```

```python
x
type(x)
```

### 1) Number 연산
- 계산기와 마찬가지로 사칙연산 (`+`, `-`, `*`, `/`)을 이용하여 계산을 수행한다.
- 지수연산은 `**`을 이용하여 계산을 수행한다.

```python
x = 3
y = 4

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x**y)  # 3^4 = 81
```

- 나눗셈에서 몫과 나머지를 각각 구할 수 있다. 몫을 구하는 연산자는 `//`, 나머지를 구하는 연산자는 `%`이다.

```python
print(7 % 3)
print(3 % 7)
print(7 // 3)
print(3 // 7) # 0이 나옴
print(3 / 7)  # / , // 결과 차이 있음
```

### 2) String Indexing(인덱싱) 및 Slicing(슬라이싱)
- Indexing(인덱싱)은 String에서 특정 위치의 문자 하나를 가져오는 방법이다.
- ** 파이썬의 index는 0부터 시작한다. **
- Slicing(슬라이싱)은 String에서 일정 범위의 문자를 가져오는 방법이다.

```python
x = "Life is too short, You need Python"
```

- 위 문장은 첫 번째 문자는 L이고 마지막 문자는 n이고, 특수기호 및 공백까지 포함해 총 34글자로 이루어진 문장이다. ** 공백 포함! **
- 이 문장에 대해 Indexing은 다음과 같다.

```python
print(x[0])  # 실제론 1번째
print(x[4])  # 실제론 5번째  # 아무것도 없는 게 아니라 '공백'이 있음
print(x[25]) # 실제론 26번째
```

- x[0]은 첫 번째 문자 `L`을 의미한다.
- x[4]는 다섯 번째 문자인 공백을 의미한다.
- x[25]는 26번째 문자 `e`를 의미한다.

```python
print(x[-1]) # 0부터 시작해서 뒤로 1번째
print(x[-5]) # 뒤에서 5번째
```

- 파이썬에서는 `-`를 이용하여 Indexing할 수 있다.
- `x[-1]`은 뒤에서 세어 첫 번째라는 것을 의미하기 때문에 `n`을 추출하고, `x[-5]`는 뒤에서 다섯 번째인 `y`를 추출하게 된다.
- Slicing은 숫자로 범위를 지정해주어 String에서 일정 범위의 문자를 추출하는 역할을 한다. 이때, `:` 기호를 이용한다.

```python
x[0:4] # 0부터 4직전까지
```

- `x[0:4]`는 index `0`부터 `4` 직전까지의 문자를 가져온다.
- 따라서 index 0, 1, 2, 3에 해당하는 `Life`가 추출된다.
- Slicing에서는 마지막 index가 포함되지 않는다.

```python
x[5:7] # 5부터 7직전까지
```

- x[5:7]은 index 5부터 7 직전까지의 문자를 가져오므로 `is`가 추출된다.
- x[a:]는 index `a`부터 마지막까지 가져온다.
- x[:b]는 처음부터 index `b` 직전까지 가져온다.

```python
print(x[19:]) # 19부터 끝까지
print(x[:17]) # 처음부터 17 직전까지 **
```

- x[19:]는 index 19부터 String의 마지막까지 가져온다.
- x[:17]은 String의 처음부터 index 17 직전까지 가져온다.

### 3) String 관련 method
- `method`는 특정 자료형에서 사용할 수 있도록 정의된 기능이다.
- ** String method는 String 뒤에 `.`을 붙여 사용한다. **
- 자주 사용하는 String method 몇 가지를 살펴본다.

#### 1. count(x)
- String에서 특정 문자 `x`가 나타나는 개수를 반환한다.

```python
x = "hobby"
x.count('b')
```

```python
x = "hobby"
x.count('a')  # 아무것도 없을 때 0 반환 # NA 나오는 경우는 뒤에 있음
```

- ** `count()` method는 대소문자를 구분한다. ***

#### 2. find(x), index(x)
- String에서 특정 문자 `x`의 위치를 찾아준다.

```python
x = "Python is the best choice"
print(x.find('i'))
print(x.find('k'))  # find는 없으면 -1 출력 # 결과 나중에 오류날 수 있음 # 그냥 코드 돌아가게 만들고 싶을 때 사용 ******
print(x.index('b'))
print(x.index('k')) # index를 사용하면 찾는 문자 없을 경우 에러남 # 에러 찾고 싶을 때 사용 ******
```

- `find()` method는 찾고자 하는 문자가 없다면 `-1`을 반환하고, `index()` method는 오류가 발생한다.
- 두 method 모두 대소문자를 구분한다.

#### 3. upper(), lower()
- `upper()`는 영문자를 대문자로, `lower()`는 영문자를 소문자로 변환한다.

```python
x = "hi"
print(x.upper())

x = "HI"
print(x.lower())
```

```python
## +)
x = "hiHI"
print(x.upper())
print(x.lower())
```

#### 4. strip() *
- String의 양쪽에 있는 공백을 제거한다.

```python
x = " hi "

x.strip()

x = "hi hi" # 중간 공백은 못 없앰
x.strip()
```

#### 5. replace(x, y)
- String에서 `x`에 해당하는 부분을 `y`로 변경한 새로운 String을 반환한다.

```python
x = "Life is too short"
x.replace("Life", "My arm")
```

#### 6. split(x)
- String을 지정한 구분자를 기준으로 나눈다. # *꼭 default 알아야 함!!*
- 구분자를 지정하지 않으면 공백을 기준으로 String을 나눈다.
- 나눈 결과는 List(리스트) 형태로 반환된다.

```python
x.split()
```
