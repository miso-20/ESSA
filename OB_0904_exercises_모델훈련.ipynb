## **모델 훈련 연습 문제**
___

- 개념 문제의 경우 텍스트 셀을 추가하여 정답을 적어주세요.

### **1. 수백만 개의 특성을 가진 훈련 세트에서는 어떤 선형 회귀 알고리즘을 사용할 수 있을까요?**
___


수백만 개의 특성이 있다면 계산 복잡도가 매우 높은 정규방정식이나 SVD(특잇값 분해) 방식은 사용할 수 없다. 따라서 확률적 경사 하강법(SGD)이나 미니배치 경사 하강법을 사용해야 한다. 훈련 세트가 메모리에 다 들어간다면 배치 경사 하강법도 사용할 수 있다.

### **2. 배치 경사 하강법을 사용하고 에포크마다 검증 오차를 그래프로 나타내봤습니다. 검증 오차가 일정하게 상승되고 있다면 어떤 일이 일어나고 있는 걸까요? 이 문제를 어떻게 해결할 수 있나요?**
___

두 가지 가능성이 있다.
- 만약 훈련 오차도 함께 상승하고 있다면, 학습률(learning rate)이 너무 높아 알고리즘이 발산(diverging)하고 있는 것이다. 이 경우 학습률을 낮춰야 한다.

- 훈련 오차는 감소하는데 검증 오차만 상승하고 있다면, 모델이 훈련 데이터에 과대적합(overfitting)되고 있는 것이다. 이 경우 훈련을 조기 종료(Early Stopping)하거나, 릿지/라쏘 같은 규제를 추가하거나, 더 많은 훈련 데이터를 수집하여 해결할 수 있다.

### **3. 릿지 회귀를 사용했을 때 훈련 오차가 검증 오차가 거의 비슷하고 둘 다 높았습니다. 이 모델에는 높은 편향이 문제인가요, 아니면 높은 분산이 문제인가요? 규제 하이퍼파라미터 $\alpha$를 증가시켜야 할까요 아니면 줄여야 할까요?**
___

훈련 오차와 검증 오차가 비슷하면서 둘 다 높다는 것은 모델이 훈련 데이터에 과소적합(underfitting)되었다는 뜻이다. 따라서 이 모델은 높은 편향(High Bias)을 가진 것이 문제이다. 과소적합을 해결하려면 모델을 더 유연하게(복잡하게) 만들어야 하므로, 규제 하이퍼파라미터 α를 줄여야 한다.

### **4. 다음과 같이 사용해야 하는 이유는?**
___
- 평범한 선형 회귀(즉, 아무런 규제가 없는 모델) 대신 릿지 회귀
- 릿지 회귀 대신 라쏘 회귀
- 라쏘 회귀 대신 엘라스틱넷


-> 평범한 선형 회귀 대신 릿지 회귀: 모델에 규제가 약간이라도 있는 것이 대부분의 경우 규제가 아예 없는 것보다 성능이 좋으므로 릿지 회귀가 기본적으로 선호된다.

-> 릿지 회귀 대신 라쏘 회귀: 라쏘 회귀는 덜 중요한 특성의 가중치를 완전히 0으로 만든다. 즉, 자동으로 특성을 선택(Feature Selection)해주는 효과가 있으므로, 실제로 유용한 특성이 몇 개뿐이라고 강하게 의심될 때 사용한다.

-> 라쏘 회귀 대신 엘라스틱넷: 특성 수가 훈련 샘플 수보다 많거나, 특성들끼리 서로 강하게 연관되어 있을 때 라쏘 회귀는 불안정하게 작동할 수 있다. 이럴 때는 릿지와 라쏘를 혼합한 엘라스틱넷이 훨씬 안정적이므로 더 선호된다.
